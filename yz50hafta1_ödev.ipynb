# **SORU 3**

```python
prediction = [0.8,0.4,0.5,0.6,0.45]
target = [0.6,0.4,0.34,0.42,0.56]

def loss(prediction,target):

  cost = 0
  for i in range(len(target)):

    cost += (target[i] - prediction[i])**2 / len(target)
  return cost

result = loss(prediction,target)
print(result)
```
